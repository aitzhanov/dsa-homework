````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).

````python
# База
cities_clean = [city.strip().lower() for city in cities] 
cities_clean
````

````python
count = 0
amount = 0

for i in range (len(paid)):
    if not paid[i]:
        count += 1
        amount += amounts_int[i]

print(f"Total count of unpaid orders: {count}, Total amount of unpaid orders: {amount}")
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# Стандарт
amount_astana = 0
amount_almaty = 0

for i in range (len (paid) ):
    if paid[i]:
        if cities_clean[i] == 'astana':
            amount_astana += amounts_int[i]
        if cities_clean[i] == 'almaty':
            amount_almaty += amounts_int[i]

print ("Amount of paid orders in Astana:", amount_astana)
print ("Amount of paid orders in Almaty:", amount_almaty)
````

````python
biggest = 0
city = ''
index = 0

for i in range (len (amounts_int) ):
    if amounts_int [i] >= biggest and paid[i]:
        biggest = amounts_int [i]
        city = cities_clean [i]
        index = i

print (f"Biggest order is {biggest} from {city} and index is {index}")
````

````python
count_weekend = 0

for day in days:
    if day == "сб" or day == "вс":
        count_weekend += 1

percentage = count_weekend / len (days) * 100

print(f"Weekend orders: {count_weekend}")
print(f"Percentage: {int(percentage)}%")
````

````python
summ = 0
months = 0

while (summ <= 400000):
    summ += 15000
    months += 1

print (months)

#Каждый месяц на 1000 тенге больше чем в прошлый

summ = 0
amount = 15000
months = 0

while (summ <= 400000):
    summ += amount
    amount += 1000
    months += 1

print (months)
````

13. *Подумать:* какие проблемы данных вы встретили (пробелы, регистр, числа-строки)? Опишите 2–3 предложениями в Markdown-ячейке. Именно это мы будем делать в Pandas на занятии 6.

````python
# 10
unique_cities = []

for city in cities_clean:
    if city not in unique_cities:
        unique_cities.append(city)

print(unique_cities)

revenue_by_city = []

for city in unique_cities:
    revenue = 0

    for i in range (len (cities_clean)):
        if cities_clean[i] == city and paid[i]:
            revenue += amounts_int[i]

    revenue_by_city.append(revenue)

print(revenue_by_city)
````

````python
# 11
median = 0
sorted_amounts = sorted (amounts_int)

if len (sorted_amounts) % 2 == 0:
    median = ( sorted_amounts [len(sorted_amounts) // 2] + sorted_amounts [(len(sorted_amounts) // 2) - 1] ) / 2 #потому что индекс начинается с 0
if len (sorted_amounts) % 2 == 1:
    median = sorted_amounts [(len(sorted_amounts) // 2)]

median
````

````python
#12
weekdays = ["пн", "вт", "ср", "чт", "пт", "сб", "вс"]

for day in weekdays:
    revenue = 0

    for i in range(len(days)):
        if days[i] == day and paid[i]:
            revenue += amounts_int[i]

    blocks = revenue // 5000

    print(f"{day} | {'█' * blocks}  {revenue}")
````
